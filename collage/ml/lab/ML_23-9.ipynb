```python
import numpy as np
import pandas as pn
```

```python
data = pn.read_csv("Housing.csv")
data
```

```python
data.isnull().sum()
```

```python
old_data = data.copy()
```

```python
new_data = pn.get_dummies(data)
new_data
```

```python
new_data['price'].describe().astype(int)
```

```python
new_data['new_price'] = np.where(new_data['price']<=3430000,'C1',
                                 np.where(new_data['price']<=4340000,'C2',
                                 np.where(new_data['price']<=5740000,'C3','C4')))
```

```python
new_data
```

```python
new_data.drop(['price'],axis=1,inplace=True)
```

```python
new_data.corr()
```

```python
y = new_data.iloc[:,-1]
x = new_data.iloc[:,:-1]
```

```python
x.shape
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.30)
```

```python
from sklearn.tree import DecisionTreeClassifier
dtc = DecisionTreeClassifier(criterion='entropy')
dtc.fit(x_train,y_train)
```

```python
y_pred = dtc.predict(x_test)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```

```python
from sklearn.metrics import precision_score
print(precision_score(y_test,y_pred,average="macro"))
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test,y_pred))
```
